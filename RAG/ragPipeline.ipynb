DATA INGESTION


```python
#TEXT LOADER

from langchain_community.document_loaders import TextLoader

loader = TextLoader("intro.txt")
doc = loader.load()
print(doc[0].page_content)
```

```python
#WEB BASED LOADER

from langchain_community.document_loaders import WebBaseLoader
import bs4

#Load, chunk, index the content of a webpage

web_path = "https://www.geeksforgeeks.org/python/python-programming-language-tutorial/"

# loader = WebBaseLoader(web_path)
# doc = loader.load()
# print(doc[0].page_content[:500])  

loader = WebBaseLoader(web_path, bs_kwargs=dict(parse_only=bs4.SoupStrainer(class_=("code_output","mainSubHeaderDiv"))))
loader = WebBaseLoader(web_path)
doc = loader.load()
print(doc)
```

```python
#PDF Loader

from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("appointment.PDF")
doc = loader.load()
doc
```

CHUNKING

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

chunking = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
documents = chunking.split_documents(doc)
documents[:5]
```

EMBEDDING

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
db = Chroma.from_documents(documents[:20],embeddings)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
db = FAISS.from_documents(documents[:20], embeddings)
```

```python
query = "how much is stipend?"
result = db.similarity_search(query)
result[0].page_content
```

CHAIN AND RETRIEVER 

```python
from langchain_google_genai import ChatGoogleGenerativeAI
import os
    
llm = ChatGoogleGenerativeAI(model="gemini-3.7-flash",api_key=os.getenv("GEMMA_APIKEY"))
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
"Use the following pieces of context to answer the question at the end. If you don't know the answer, just say that you don't know, don't try to make up an answer." \
"<context> {context} </context> \n\n Question: {input} \n Answer:"
)
```

```python
#Create stuff document chains

from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_classic.chains import create_retrieval_chain

document_chain = create_stuff_documents_chain(llm,prompt)

retriever = db.as_retriever()

retrieval_chain = create_retrieval_chain(retriever,document_chain)
```

```python
result = retrieval_chain.invoke({
    "context": "Some retrieved docs go here",
    "input": "how much is stipend?"
})
result
```
